```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from tensorflow.keras.utils import to_categorical
from sklearn import datasets
```

```python
iris = datasets.load_iris()
X = iris.data
target = iris.target
Y = to_categorical(target).astype("uint8")
# 0 => [1 0 0]
# 1 => [0 1 0]
# 2 => [0 0 1]
```

```python
idx = 0
print(X[idx])
print(target[idx])
print(Y[idx])
```

```python
model = keras.Sequential(
  [
    layers.Dense(3, activation="sigmoid", name="layer1", input_shape=(4,)),
    layers.Dense(3, activation="sigmoid", name="layer2"),
  ]
)

model.summary()
```

```python
y = model(X)
print(y[149])
```

```python
optimizer=tf.keras.optimizers.SGD(learning_rate=0.03)
loss=tf.keras.losses.MeanSquaredError()

model.compile(optimizer, loss)
history = model.fit(X, Y, epochs=500)
```

```python
plt.plot(history.history['loss'])
plt.show()
```

```python
pd.DataFrame(history.history).plot(figsize=(8,5))
plt.show()
```

```python
model.predict(np.array([[3,3,1,1]]))
```

##RED PROFUNDA



```python
model_profunda = keras.Sequential(
  [
    layers.Dense(3, activation="sigmoid", name="layer1", input_shape=(4,)),
    layers.Dense(3, activation="sigmoid", name="layer2"),
    layers.Dense(3, activation="sigmoid", name="layer3"),
    layers.Dense(3, activation="sigmoid", name="layer4"),  # salida = 3 neuronas
  ]
)

model_profunda.summary()
```

```python
model_profunda.compile(
    optimizer=tf.keras.optimizers.SGD(learning_rate=0.03),
    loss=tf.keras.losses.MeanSquaredError(),
)

history_deep = model_profunda.fit(X, Y, epochs=500)
```

```python
plt.plot(history.history["loss"], label="Original")
plt.plot(history_deep.history["loss"], label="Tres capas ocultas")
plt.title("Comparación de pérdida")
plt.xlabel("Épocas")
plt.ylabel("Loss")
plt.legend()
plt.show()

print("Loss final original:", history.history["loss"][-1])
print("Loss final profundo:", history_deep.history["loss"][-1])
```

```python
model_profunda.predict(np.array([[3, 3, 1, 1]]))
```

```python
model_profunda = keras.Sequential(
  [
    layers.Dense(3, activation="relu", name="layer1", input_shape=(4,)),
    layers.Dense(3, activation="relu", name="layer2"),
    layers.Dense(3, activation="relu", name="layer3"),
    layers.Dense(3, activation="softmax", name="layer4"),  # salida = 3 neuronas
  ]
)

model_profunda.summary()
```

```python
model_profunda.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.005),
    loss=tf.keras.losses.CategoricalCrossentropy(),
)

history_deep = model_profunda.fit(X, Y, epochs=500)
```

```python
plt.plot(history.history["loss"], label="Original")
plt.plot(history_deep.history["loss"], label="Tres capas ocultas")
plt.title("Comparación de pérdida")
plt.xlabel("Épocas")
plt.ylabel("Loss")
plt.legend()
plt.show()

print("Loss final original:", history.history["loss"][-1])
print("Loss final profundo:", history_deep.history["loss"][-1])
```
